```python
import fitz
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
page = doc[3]
```

```python
pix = page.get_pixmap(matrix=fitz.Matrix(4,4))

pix.save("page.png")
```

```python
import os

print(os.listdir())
```

```python
img = cv2.imread("page.png")
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(12,10))
plt.imshow(img_rgb)
plt.axis("off")
plt.show()
```

```python
crop = img[700:1700, 250:2200]
```

```python
crop_rgb = cv2.cvtColor(crop, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(14,8))
plt.imshow(crop_rgb)
plt.axis("off")
plt.show()
```

```python
hsv = cv2.cvtColor(crop, cv2.COLOR_BGR2HSV)
```

```python
lower_blue = np.array([80, 50, 50])
upper_blue = np.array([130, 255, 255])

mask = cv2.inRange(hsv, lower_blue, upper_blue)
```

```python
plt.figure(figsize=(14,8))
plt.imshow(mask, cmap="gray")
plt.axis("off")
plt.show()
```

```python
num_labels, labels, stats, centroids = (
    cv2.connectedComponentsWithStats(mask)
)
```

```python
dots = []

for i in range(1, num_labels):

    area = stats[i, cv2.CC_STAT_AREA]

    if 10 < area < 300:

        cx, cy = centroids[i]

        dots.append((cx, cy))

print("Dots detected:", len(dots))
```

```python
test = crop.copy()

for cx, cy in dots:

    cv2.circle(
        test,
        (int(cx), int(cy)),
        10,
        (0,255,0),
        3
    )

test_rgb = cv2.cvtColor(test, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(14,8))
plt.imshow(test_rgb)
plt.axis("off")
plt.show()
```

```python
top_y = 100
bottom_y = 850

top_rate = 5.0
bottom_rate = 2.5
```

```python
def y_to_rate(y):

    rate = top_rate - (
        (y - top_y)
        / (bottom_y - top_y)
    ) * (top_rate - bottom_rate)

    rounded = round(rate / 0.125) * 0.125

    return rounded
```

```python
for i in range(5):

    print(
        dots[i],
        "→",
        y_to_rate(dots[i][1])
    )
```

```python
from sklearn.cluster import KMeans
```

```python
x_values = np.array(
    [d[0] for d in dots]
).reshape(-1,1)

kmeans = KMeans(
    n_clusters=4,
    random_state=0,
    n_init=10
)

kmeans.fit(x_values)

centers = sorted(
    kmeans.cluster_centers_.flatten()
)

print(centers)
```

```python
rows = []

for cx, cy in dots:

    nearest = min(
        centers,
        key=lambda c: abs(cx-c)
    )

    if nearest == centers[0]:
        year = "2026"

    elif nearest == centers[1]:
        year = "2027"

    elif nearest == centers[2]:
        year = "2028"

    else:
        year = "Longer Run"

    rows.append({
        "Year": year,
        "Projection": y_to_rate(cy)
    })

df = pd.DataFrame(rows)

print(df.head())
```

```python
df.to_csv(
    "fomc_projections.csv",
    index=False
)

print("CSV saved successfully")
```
