```python
import fitz
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
```

```python
# PDF INFO (change as needed)
pdf_path = "Summary of Economic Projections, March 18, 2026.pdf"
starting_year = 2026

# KMEANS (change as needed based on x-axis)
n_clusters = 4


# GRAPH CROP
top = 410
bottom = 1750
left = 320
right = 2200

# GRID CALIBRATION
top_grid_y = 28
top_grid_value = 6.0

pixels_per_0125 = 22.5
```

```python
def detect_dots(mask):

    num_labels, labels, stats, centroids = (
        cv2.connectedComponentsWithStats(
            mask,
            connectivity=8
        )
    )

    centers = []

    for i in range(1, num_labels):

        area = stats[i, cv2.CC_STAT_AREA]

        if area > 20:

            cx, cy = centroids[i]

            centers.append((cx, cy))

    return centers
```

```python
pdf_path = pdf_path
doc = fitz.open(pdf_path)

print("Pages:", len(doc))
```

```python
page = doc[3]
```

```python
pix = page.get_pixmap(matrix=fitz.Matrix(4,4))

pix.save("page.png")
```

```python
img = cv2.imread("page.png")
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
plt.figure(figsize=(20,20))
plt.imshow(img_rgb)
plt.axis("on")
plt.show()
```

```python
crop = img[top:bottom, left:right]
```

```python
crop_rgb = cv2.cvtColor(crop, cv2.COLOR_BGR2RGB)
```

```python
plt.figure(figsize=(18,12))
plt.imshow(crop_rgb)
plt.axis("on")
plt.show()
```

```python
gray = cv2.cvtColor(crop, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(14,8))
plt.imshow(gray, cmap="gray")
plt.axis("off")
plt.show()
```

```python
edges = cv2.Canny(
    gray,
    50,
    150
)

plt.figure(figsize=(14,8))
plt.imshow(edges, cmap="gray")
plt.axis("off")
plt.show()
```

```python
lines = cv2.HoughLinesP(
    edges,
    rho=1,
    theta=np.pi/180,
    threshold=100,
    minLineLength=1000,
    maxLineGap=20
)
```

```python
line_img = crop.copy()

if lines is not None:

    for line in lines:

        x1, y1, x2, y2 = line[0]

        if abs(y2 - y1) < 5:

            cv2.line(
                line_img,
                (x1, y1),
                (x2, y2),
                (0,255,0),
                3
            )

line_rgb = cv2.cvtColor(
    line_img,
    cv2.COLOR_BGR2RGB
)

plt.figure(figsize=(18,12))
plt.imshow(line_rgb)
plt.axis("off")
plt.show()
```

```python
y_lines = []

if lines is not None:

    for line in lines:

        x1, y1, x2, y2 = line[0]

        if abs(y2 - y1) < 5:

            y_lines.append(y1)

print(y_lines[:50])
```

```python
unique_y = []

for y in sorted(y_lines):

    if not unique_y:

        unique_y.append(y)

    elif abs(y - unique_y[-1]) > 10:

        unique_y.append(y)

print(unique_y)
```

```python
line_view = crop.copy()

for y in unique_y:

    cv2.line(
        line_view,
        (0, int(y)),
        (line_view.shape[1], int(y)),
        (255,0,0),
        2
    )

line_view_rgb = cv2.cvtColor(
    line_view,
    cv2.COLOR_BGR2RGB
)

plt.figure(figsize=(18,12))
plt.imshow(line_view_rgb)
plt.axis("off")
plt.show()
```

```python
line_labels = crop.copy()

for y in unique_y:

    cv2.line(
        line_labels,
        (0, int(y)),
        (line_labels.shape[1], int(y)),
        (255,0,0),
        2
    )

    cv2.putText(
        line_labels,
        str(int(y)),
        (50, int(y)-5),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (255,0,0),
        2
    )

line_labels_rgb = cv2.cvtColor(
    line_labels,
    cv2.COLOR_BGR2RGB
)

plt.figure(figsize=(20,14))
plt.imshow(line_labels_rgb)
plt.axis("off")
plt.show()
```

```python
hsv = cv2.cvtColor(crop, cv2.COLOR_BGR2HSV)
```

```python
lower_blue = np.array([80, 50, 50])
upper_blue = np.array([130, 255, 255])

mask = cv2.inRange(hsv, lower_blue, upper_blue)
```

```python
plt.figure(figsize=(14,8))
plt.imshow(mask, cmap="gray")
plt.axis("off")
plt.show()
```

```python
kernel = np.ones((3,3), np.uint8)

mask_clean = cv2.morphologyEx(
    mask,
    cv2.MORPH_OPEN,
    kernel
)

mask_clean = cv2.morphologyEx(
    mask_clean,
    cv2.MORPH_CLOSE,
    kernel
)
```

```python
plt.figure(figsize=(14,8))
plt.imshow(mask_clean, cmap="gray")
plt.axis("off")
plt.show()
```

```python
centers = detect_dots(mask_clean)
```

```python
print(centers[:10])
```

```python
test = crop.copy()

for cx, cy in centers:

    cv2.circle(
        test,
        (int(cx), int(cy)),
        10,
        (0,255,0),
        3
    )

test_rgb = cv2.cvtColor(test, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(18,12))
plt.imshow(test_rgb)
plt.axis("off")
plt.show()
```

```python
x_values = np.array(
    [d[0] for d in centers]
).reshape(-1,1)
```

```python
kmeans = KMeans(
    n_clusters=n_clusters,
    random_state=0,
    n_init=10
)

kmeans.fit(x_values)
```

```python
x_cluster_centers = list(sorted(
    kmeans.cluster_centers_.flatten()
))
print(centers)
```

```python
top_y = 20
bottom_y = 1040

top_rate = 4.5
bottom_rate = 2.0
```

```python
def y_to_rate(y):

    steps = (
        (y - top_grid_y)
        / pixels_per_0125
    )

    rate = (
        top_grid_value
        - (steps * 0.125)
    )

    rounded = (
        round(rate / 0.125)
        * 0.125
    )

    return rounded
```

```python
rows = []

for cx, cy in centers:

    nearest = min(
        x_cluster_centers,
        key=lambda c: abs(cx - c)
    )

    cluster_index = (
        x_cluster_centers.index(nearest)
    )

    if cluster_index < n_clusters - 1:

        year = str(
            starting_year
            + cluster_index
        )

    else:

        year = "Longer Run"

    rows.append({
        "Year": year,
        "Projection": y_to_rate(cy)
    })

df = pd.DataFrame(rows)
```

```python
print(df.sort_values(["Year", "Projection"]))
```

```python
print(df.groupby("Year").size())
```

```python
df_sorted = df.sort_values(
    ["Year", "Projection"]
)

df_sorted.to_csv(
    "projections.csv",
    index=False
)

print("CSV saved")
```
